```python
import pandas as pd
import numpy as np
```

```python
dates = pd.date_range('1/1/2000', periods=8)
df = pd.DataFrame(np.random.randn(8, 4),
                  index=dates, columns=['A', 'B', 'C', 'D'])
```

```python
df
```

### ".loc"
With the `.loc[]` command you can provide the **names** of rows and columns, to access data. There are various combinations possible on how to do that:

1) Single entry for row and column
2) List of individual entries (can be put in either or both of the row and column locator)
3) Range from Start to End column/row name (this includes the actually defined end - different to index ranges!)

***See examples below!***

```python
df.loc["2000-01-04", "B"]
```

```python
df.loc[["2000-01-01", "2000-01-04"], "B"]
```

```python
df.loc[["2000-01-01", "2000-01-04"], "B":"D"]
```

### ".iloc"
With the `.iloc[]` command you can provide the **indices** of rows and columns, to access data. There are various combinations possible on how to do that:

1) Single entry for row and column
2) All rows/columns using the `:` argument
3) List of individual entries (can be put in either or both of the row and column locator)
4) Range from Start to ***End+1*** column/row indices (this ***does NOT include the actually defined end*** - different to `.loc` ranges!)

***The examples below show the exact same output as before using the `.loc` command.***

```python
df.iloc[3,1]

# you can see for yourself by comparing element-wise if the output is identical:
# df.loc["2000-01-04", "B"] == df.iloc[3,1]
```

```python
df.iloc[[0,3],1]
# df.loc[["2000-01-01", "2000-01-04"], "B"] == df.iloc[[0,3],1]
```

```python
df.iloc[[0,3],1:4]
# df.loc[["2000-01-01", "2000-01-04"], "B":"D"] == df.iloc[[0,3],1:4]
```
