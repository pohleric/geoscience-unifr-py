```python
# creating random "heights" in a matrix (2D) with 6 rows and 5 columns
import numpy as np

# 30 random values with mean=170, sd=10, n=30:
r30 = np.random.normal(170,10,30)

# assigning those 30 values to a
# 6 x 5 matrix (.reshape(rows,columns)) 
# and transform the values to integer type (.astype(int))
cr = np.array(r30).reshape((6,5)).astype(int)
```

```python
cr
```

```python
# number of rows and columns (2 values as tuple) stored in the matrix parameter called "shape"
cr.shape
```

```python
# save as two variables (separated with ",")
rows, cols = cr.shape
```

```python
# going through each entry, row by row, and column by column
min_value = cr[0,0]
for h in range(cols):
    # example first row (cr[0,h]) << 0th row 
    print(cr[0,h])
```

```python
# going through each entry, row by row, and column by column
min_value = cr[0,0]
for c in range(cols):
    # example first row (cr[0,c]) << 0th row 
    # whenever cr[0,h] is smaller than min_value, overwrite the value of min_value with the new value
    if cr[0,c] < min_value:
        min_value = cr[0,c]
print(min_value)
```

```python
# do the same but iterate also through all the rows (> cr[r,c]
min_value = cr[0,0]
for r in range(rows):
    for c in range(cols):
        # print(cr[r,c])
        if cr[r,c] < min_value:
            min_value = cr[r,c]
print(min_value)
```

```python
# getting the positions (indices: row and column) instead of the actual value:
min_value = cr[0,0]
min_value_indices = [0,0] # < list with two entries for inices of row and column
for r in range(rows):
    for c in range(cols):
        # print(cr[r,c])
        if cr[r,c] < min_value:
            min_value = cr[r,c]
            min_value_indices = [r,c] # only row and column indices
print(min_value)
print(min_value_indices)
```

```python
# check if the indices do actually give the minimum number:
cr[min_value_indices[0], min_value_indices[1]]
```

```python
# Attention: the variable "min_value_indices" is a list
# If directly put in as index into "cr" it will yield rows 1 and 4:
# cr[[1,4]]
cr[min_value_indices]
```

```python
cr[[1,4],]
```

### Indices over values
Why getting the indices can be advantageous over finding just the smalles value?

```python
# another table (matrix) that stores another variable, for example the person's ID (here A1:A30)
IDs = range(30)
IDs = ["A"+ str(i+1) for i in IDs]
ids = np.array(IDs).reshape(6,5)
ids
```

```python
# which person ID corresponds to the smallest identified person?

# Using the indices to find the smalles value:
cr[min_value_indices[0], min_value_indices[1]]
```

```python
# Using the same indices but now in the table of person IDs
ids[min_value_indices[0], min_value_indices[1]]
```
